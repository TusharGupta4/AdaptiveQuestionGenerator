```python
# ============================================================
# UPLOAD AND AUTOMATICALLY DETECT THE PDF
# ============================================================

!pip install -q pymupdf

import fitz
import os
from google.colab import files

# Upload PDF
uploaded = files.upload()

# Check what was actually uploaded
print("=" * 80)
print("UPLOADED FILES")
print("=" * 80)

for filename in uploaded.keys():
    print("✓", filename)

# ------------------------------------------------------------
# Automatically find the uploaded PDF
# ------------------------------------------------------------

pdf_files = [
    filename
    for filename in uploaded.keys()
    if filename.lower().endswith(".pdf")
]

if len(pdf_files) == 0:
    raise FileNotFoundError("No PDF file was uploaded.")

if len(pdf_files) > 1:
    print("\nMultiple PDFs found:")
    for i, filename in enumerate(pdf_files):
        print(i, filename)

    # Use the first PDF
    pdf_file = pdf_files[0]
else:
    pdf_file = pdf_files[0]

print("\nUsing PDF:")
print(pdf_file)

# ------------------------------------------------------------
# Check that the file exists
# ------------------------------------------------------------

if not os.path.exists(pdf_file):
    raise FileNotFoundError(
        f"Uploaded PDF could not be found: {pdf_file}"
    )

# ------------------------------------------------------------
# Open PDF
# ------------------------------------------------------------

doc = fitz.open(pdf_file)

print("\n" + "=" * 80)
print("PDF INFORMATION")
print("=" * 80)

print("File:", pdf_file)
print("Total pages:", len(doc))

# ------------------------------------------------------------
# Extract text page by page
# ------------------------------------------------------------

page_texts = []

for page_number, page in enumerate(doc, start=1):

    text = page.get_text("text")

    page_texts.append(text)

    print(
        f"Page {page_number:02d}: "
        f"{len(text):5d} characters"
    )

# ------------------------------------------------------------
# Separate question paper and answer key
# ------------------------------------------------------------

question_paper_text = "\n".join(page_texts[:7])

answer_key_text = "\n".join(page_texts[7:])

# ------------------------------------------------------------
# Basic validation
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("VALIDATION")
print("=" * 80)

print("Question paper pages:", min(7, len(page_texts)))
print("Answer key pages:", max(0, len(page_texts) - 7))

if "SECTION A" in question_paper_text:
    print("✓ Section A found")
else:
    print("⚠ Section A not found")

if "SECTION B" in question_paper_text:
    print("✓ Section B found")
else:
    print("⚠ Section B not found")

if "Question 1" in question_paper_text:
    print("✓ Question 1 found")

if "Question 5" in question_paper_text:
    print("✓ Question 5 found")

if "ANSWER KEY" in answer_key_text:
    print("✓ Answer key found")
else:
    print("⚠ Answer key heading not detected")

print("\n✓ PDF extraction completed successfully.")
```

```python
# ============================================================
# ISC CLASS 12 ENGLISH PAPER 2 — DATABASE-READY DATASET
# ============================================================

import pandas as pd
import re

# ------------------------------------------------------------
# 1. CREATE A COPY
# ------------------------------------------------------------

df_db = df_final.copy()

# ------------------------------------------------------------
# 2. BASIC PAPER METADATA
# ------------------------------------------------------------

df_db["board"] = "ISC"
df_db["class"] = "Class 12"
df_db["subject"] = "English"
df_db["year"] = 2025
df_db["paper"] = "Paper 2 - Literature in English"
df_db["source_type"] = "Specimen Question Paper"
df_db["source_file"] = "ENGLISH-PAPER-2 (1).pdf"

# ------------------------------------------------------------
# 3. CREATE QUESTION ID
# ------------------------------------------------------------

# Example:
# Q1(i)       -> ISC12_ENG2_2025_Q1_i
# Q3(iii)(a)  -> ISC12_ENG2_2025_Q3_iii_a

def create_question_id(qno):

    clean = qno

    # Remove brackets
    clean = clean.replace("(", "_")
    clean = clean.replace(")", "")

    # Remove any unwanted spaces
    clean = clean.replace(" ", "")

    return f"ISC12_ENG2_2025_{clean}"


df_db["question_id"] = (
    df_db["question_number"]
    .apply(create_question_id)
)

# ------------------------------------------------------------
# 4. EXTRACT PARENT QUESTION
# ------------------------------------------------------------

# Q1(i)       -> Q1
# Q3(iii)(a)  -> Q3
# Q5(iii)(b)  -> Q5

df_db["parent_question"] = (
    df_db["question_number"]
    .str.extract(r"(Q[1-5])")[0]
)

# ------------------------------------------------------------
# 5. EXTRACT SUB-QUESTION
# ------------------------------------------------------------

def get_sub_question(qno):

    match = re.search(r"Q[1-5](.*)", qno)

    if match:
        sub = match.group(1)

        if sub == "":
            return None

        # Convert:
        # (i)       -> i
        # (iii)(a)  -> iii-a
        # (iii)(b)  -> iii-b

        sub = sub.replace("(", "")
        sub = sub.replace(")", "")
        sub = sub.replace("-", "")

        # Need to preserve hierarchy
        parts = re.findall(r"[ivx]+|[a-z]+", sub)

        if len(parts) > 1:
            return "-".join(parts)

        return parts[0] if parts else None

    return None


df_db["sub_question"] = (
    df_db["question_number"]
    .apply(get_sub_question)
)

# ------------------------------------------------------------
# 6. INSTRUCTION FIELD
# ------------------------------------------------------------

# The current extracted dataset already contains the complete
# question wording, including word-limit instructions where
# they were captured as part of the question.
#
# We are not inventing additional instructions.

df_db["instruction"] = None

# ------------------------------------------------------------
# 7. KEEP ONLY THE 16 MYSQL COLUMNS
# ------------------------------------------------------------

mysql_columns = [
    "question_id",
    "board",
    "class",
    "subject",
    "year",
    "paper",
    "source_type",
    "parent_question",
    "sub_question",
    "question_type",
    "marks",
    "question_text",
    "answer",
    "source_file",
    "question_number",
    "instruction"
]

df_db = df_db[mysql_columns]

# ------------------------------------------------------------
# 8. VALIDATION — COLUMN STRUCTURE
# ------------------------------------------------------------

print("=" * 80)
print("DATABASE-READY DATASET VALIDATION")
print("=" * 80)

print("\nNumber of records:", len(df_db))

print("\nNumber of columns:", len(df_db.columns))

print("\nColumns:")
for i, col in enumerate(df_db.columns, start=1):
    print(f"{i}. {col}")

# ------------------------------------------------------------
# 9. CHECK REQUIRED COLUMNS
# ------------------------------------------------------------

expected_columns = [
    "question_id",
    "board",
    "class",
    "subject",
    "year",
    "paper",
    "source_type",
    "parent_question",
    "sub_question",
    "question_type",
    "marks",
    "question_text",
    "answer",
    "source_file",
    "question_number",
    "instruction"
]

if list(df_db.columns) == expected_columns:
    print("\n✓ Column structure exactly matches MySQL table.")
else:
    print("\n⚠ Column structure does NOT match expected schema.")

# ------------------------------------------------------------
# 10. DUPLICATE QUESTION ID CHECK
# ------------------------------------------------------------

duplicate_ids = df_db[
    df_db["question_id"].duplicated(keep=False)
]

print("\nDuplicate question IDs:",
      len(duplicate_ids))

if len(duplicate_ids) == 0:
    print("✓ No duplicate question IDs.")
else:
    print("⚠ Duplicate question IDs found.")

# ------------------------------------------------------------
# 11. MISSING VALUE CHECK
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("MISSING VALUE CHECK")
print("=" * 80)

for col in [
    "question_id",
    "board",
    "class",
    "subject",
    "year",
    "paper",
    "source_type",
    "parent_question",
    "question_type",
    "marks",
    "question_text",
    "answer",
    "source_file",
    "question_number"
]:

    print(
        f"{col}:",
        df_db[col].isna().sum()
    )

# ------------------------------------------------------------
# 12. CHECK QUESTION IDs
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("QUESTION ID CHECK")
print("=" * 80)

display(
    df_db[
        [
            "question_id",
            "parent_question",
            "sub_question",
            "question_number"
        ]
    ]
)

# ------------------------------------------------------------
# 13. CHECK MARKS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("MARKS CHECK")
print("=" * 80)

display(
    df_db[
        [
            "question_number",
            "marks",
            "parent_question",
            "sub_question"
        ]
    ]
)

# ------------------------------------------------------------
# 14. PAPER-LEVEL MARK VALIDATION
# ------------------------------------------------------------

actual_paper_marks = (
    df_db.groupby("parent_question")["marks"].sum()
)

print("\nStored marks by parent question:")
display(actual_paper_marks.to_frame("stored_marks"))

print("""
IMPORTANT:
The stored marks include both OR alternatives.

Actual examination marks:
Q1 = 10
Q2 = 10
Q3 = 20
Q4 = 20
Q5 = 20

Actual paper total = 80 marks.
""")

# ------------------------------------------------------------
# 15. DISPLAY FINAL DATABASE-READY DATASET
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("FINAL DATABASE-READY DATASET")
print("=" * 80)

display(df_db)

# ------------------------------------------------------------
# 16. FINAL STATUS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("FINAL STATUS")
print("=" * 80)

checks = {
    "32 records": len(df_db) == 32,
    "16 columns": len(df_db.columns) == 16,
    "Correct column order": list(df_db.columns) == expected_columns,
    "No duplicate IDs": len(duplicate_ids) == 0,
    "No missing question text":
        df_db["question_text"].isna().sum() == 0,
    "No missing answers":
        df_db["answer"].isna().sum() == 0,
    "No missing marks":
        df_db["marks"].isna().sum() == 0,
    "No missing question types":
        df_db["question_type"].isna().sum() == 0
}

for check, result in checks.items():

    if result:
        print("✓", check)
    else:
        print("⚠", check)

print("\n✓ Database-ready transformation completed.")
```

```python
# ============================================================
# GENERATE MYSQL INSERT SQL FILE — PAPER 2
# ============================================================

import pandas as pd

# ------------------------------------------------------------
# 1. MAKE SURE WE ARE USING THE DATABASE-READY DATAFRAME
# ------------------------------------------------------------

df_sql = df_db.copy()

# Convert NaN / None to SQL NULL
df_sql = df_sql.where(pd.notnull(df_sql), None)

# ------------------------------------------------------------
# 2. MYSQL STRING ESCAPING
# ------------------------------------------------------------

def mysql_value(value):

    # SQL NULL
    if value is None:
        return "NULL"

    # Handle pandas NaN
    try:
        if pd.isna(value):
            return "NULL"
    except:
        pass

    # Numeric values
    if isinstance(value, (int, float)) and not isinstance(value, bool):
        return str(value)

    # Convert to string
    value = str(value)

    # Escape backslash first
    value = value.replace("\\", "\\\\")

    # Escape single quote
    value = value.replace("'", "\\'")

    # Escape newline
    value = value.replace("\r\n", "\\n")
    value = value.replace("\n", "\\n")
    value = value.replace("\r", "\\n")

    # Escape tab
    value = value.replace("\t", "\\t")

    return f"'{value}'"


# ------------------------------------------------------------
# 3. MYSQL TABLE
# ------------------------------------------------------------

table_name = "questions"

columns = [
    "question_id",
    "board",
    "class",
    "subject",
    "year",
    "paper",
    "source_type",
    "parent_question",
    "sub_question",
    "question_type",
    "marks",
    "question_text",
    "answer",
    "source_file",
    "question_number",
    "instruction"
]

column_sql = ", ".join(columns)

# ------------------------------------------------------------
# 4. CREATE INSERT STATEMENTS
# ------------------------------------------------------------

sql_lines = []

sql_lines.append("-- =====================================================")
sql_lines.append("-- ISC CLASS 12 ENGLISH PAPER 2 — 2025")
sql_lines.append("-- DATABASE: isc_assessment")
sql_lines.append("-- TABLE: questions")
sql_lines.append("-- RECORDS: 32")
sql_lines.append("-- =====================================================")
sql_lines.append("")
sql_lines.append("USE isc_assessment;")
sql_lines.append("")

for _, row in df_sql.iterrows():

    values = [
        mysql_value(row[col])
        for col in columns
    ]

    values_sql = ", ".join(values)

    insert_statement = (
        f"INSERT INTO {table_name} ({column_sql}) "
        f"VALUES ({values_sql});"
    )

    sql_lines.append(insert_statement)

# ------------------------------------------------------------
# 5. SAVE SQL FILE
# ------------------------------------------------------------

sql_file = "ISC_Class12_English_Paper2_2025_INSERT.sql"

with open(sql_file, "w", encoding="utf-8") as f:
    f.write("\n".join(sql_lines))

print("=" * 80)
print("MYSQL SQL FILE CREATED")
print("=" * 80)

print("\nFile:", sql_file)
print("Records:", len(df_sql))
print("Columns:", len(columns))

# ------------------------------------------------------------
# 6. BASIC VALIDATION
# ------------------------------------------------------------

insert_count = sum(
    1 for line in sql_lines
    if line.startswith("INSERT INTO questions")
)

print("INSERT statements:", insert_count)

if insert_count == 32:
    print("\n✓ 32 INSERT statements generated successfully.")
else:
    print(
        f"\n⚠ Expected 32 INSERT statements, "
        f"but generated {insert_count}."
    )

print("\n✓ SQL file is ready for MySQL Workbench.")
```

```python
# ============================================================
# GENERATE ASCII-SAFE MYSQL SQL FILE — PAPER 2
# ============================================================

import pandas as pd

# ------------------------------------------------------------
# USE THE DATABASE-READY DATAFRAME
# ------------------------------------------------------------

df_sql = df_db.copy()

# ------------------------------------------------------------
# FUNCTION TO CONVERT TEXT TO UTF-8 HEX
# ------------------------------------------------------------

def mysql_utf8_hex(value):

    if value is None:
        return "NULL"

    try:
        if pd.isna(value):
            return "NULL"
    except:
        pass

    # Convert to string
    text = str(value)

    # Convert Unicode text to UTF-8 bytes
    utf8_bytes = text.encode("utf-8")

    # Convert bytes to hexadecimal
    hex_value = utf8_bytes.hex().upper()

    # CONVERT hexadecimal bytes back to UTF-8 inside MySQL
    return f"CONVERT(0x{hex_value} USING utf8mb4)"


# ------------------------------------------------------------
# FUNCTION FOR NUMERIC VALUES
# ------------------------------------------------------------

def mysql_number(value):

    if value is None:
        return "NULL"

    try:
        if pd.isna(value):
            return "NULL"
    except:
        pass

    return str(value)


# ------------------------------------------------------------
# MYSQL COLUMNS
# ------------------------------------------------------------

columns = [
    "question_id",
    "board",
    "class",
    "subject",
    "year",
    "paper",
    "source_type",
    "parent_question",
    "sub_question",
    "question_type",
    "marks",
    "question_text",
    "answer",
    "source_file",
    "question_number",
    "instruction"
]

# ------------------------------------------------------------
# GENERATE SQL
# ------------------------------------------------------------

sql_lines = []

sql_lines.append("-- ISC CLASS 12 ENGLISH PAPER 2 - 2025")
sql_lines.append("-- ASCII-SAFE UTF-8 MYSQL SCRIPT")
sql_lines.append("-- 32 RECORDS")
sql_lines.append("")
sql_lines.append("USE isc_assessment;")
sql_lines.append("")

for _, row in df_sql.iterrows():

    values = []

    for col in columns:

        if col in ["year", "marks"]:
            values.append(mysql_number(row[col]))
        else:
            values.append(mysql_utf8_hex(row[col]))

    values_sql = ", ".join(values)

    sql_lines.append(
        "INSERT INTO questions "
        "(question_id, board, class, subject, year, paper, "
        "source_type, parent_question, sub_question, "
        "question_type, marks, question_text, answer, "
        "source_file, question_number, instruction) "
        f"VALUES ({values_sql});"
    )

# ------------------------------------------------------------
# SAVE AS ASCII
# ------------------------------------------------------------

sql_file = "ISC_Class12_English_Paper2_2025_INSERT_ASCII_SAFE.sql"

with open(
    sql_file,
    "w",
    encoding="ascii"
) as f:

    f.write("\n".join(sql_lines))

# ------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------

insert_count = sum(
    1 for line in sql_lines
    if line.startswith("INSERT INTO questions")
)

print("=" * 80)
print("ASCII-SAFE MYSQL SCRIPT CREATED")
print("=" * 80)

print("\nFile:", sql_file)
print("Records:", len(df_sql))
print("INSERT statements:", insert_count)

if insert_count == 32:
    print("\n✓ 32 INSERT statements generated.")
else:
    print("\n⚠ INSERT count is incorrect.")

print("✓ File contains ASCII characters only.")
print("✓ Unicode question text is encoded as UTF-8 hexadecimal.")
print("\nReady to execute in MySQL Workbench.")
```

```python
import os
import re
import fitz
import pandas as pd

# ============================================================
# 1. Define the 7 pending unique source files
# ============================================================

pending_files = [
    "English Paper - Competency Focused Practice question.pdf",
    "English Paper - unofficial source.pdf",
    "English Paper 1 - AOPP.pdf",
    "English Paper 1 - Specimen.pdf",
    "English Paper 1 Vedantu.pdf",
    "English Paper 2 - AOPP.pdf",
    "English Paper 2 Vedantu.pdf"
]

# Check which files actually exist
available_files = []

for file_name in pending_files:
    file_path = os.path.join("/content", file_name)

    if os.path.exists(file_path):
        available_files.append(file_name)
    else:
        print(f"⚠️ File not found: {file_name}")

print("=" * 80)
print(f"Pending files found: {len(available_files)}")
print("=" * 80)

for i, file_name in enumerate(available_files, 1):
    print(f"{i}. {file_name}")


# ============================================================
# 2. Helper functions
# ============================================================

def clean_text(text):
    """Clean extracted PDF text."""
    text = text.replace("\xa0", " ")
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n\s*\n+", "\n\n", text)
    return text.strip()


def detect_document_type(file_name, full_text):
    """
    Classify the source based primarily on filename and
    explicit document terminology.
    """

    name = file_name.lower()
    text = full_text.lower()

    if "aopp" in name or "analysis of pupil performance" in text:
        return "AOPP"

    if "specimen" in name or "specimen question paper" in text:
        return "Specimen Question Paper"

    if "vedantu" in name:
        return "Third-party Question Paper"

    if "competency" in name:
        return "Competency Focused Practice"

    if "unofficial" in name:
        return "Unofficial Question Paper"

    return "Unknown"


def detect_year(text):
    """
    Detect likely paper year.
    Avoid using random four-digit numbers as year.
    """

    patterns = [
        r"\b(20(?:20|21|22|23|24|25|26))\b",
        r"year\s*[:\-]?\s*(20\d{2})",
        r"academic\s+year\s*[:\-]?\s*(20\d{2})"
    ]

    for pattern in patterns:
        match = re.search(pattern, text, re.IGNORECASE)

        if match:
            return int(match.group(1))

    return None


def detect_max_marks(text):
    """
    Detect maximum marks only when an explicit marks pattern
    exists. Do not infer from question numbers.
    """

    patterns = [
        r"maximum\s+marks\s*[:\-]?\s*(\d+)",
        r"max\.?\s*marks\s*[:\-]?\s*(\d+)",
        r"max\s*marks\s*[:\-]?\s*(\d+)",
        r"maximum\s*[:\-]?\s*(\d+)\s*marks",
        r"total\s+marks\s*[:\-]?\s*(\d+)"
    ]

    for pattern in patterns:
        match = re.search(pattern, text, re.IGNORECASE)

        if match:
            return int(match.group(1))

    return None


def detect_question_numbers(page_text):
    """
    Detect actual question numbers.

    We deliberately avoid matching years such as 2025.
    """

    questions = []

    patterns = [
        r"(?m)^\s*Question\s+([1-9]\d?)\s*[\.\):]?",
        r"(?m)^\s*Q\.?\s*([1-9]\d?)\s*[\.\):]?",
        r"(?m)^\s*([1-9])\s*[\.\)]\s+",
    ]

    for pattern in patterns:
        matches = re.findall(pattern, page_text, re.IGNORECASE)

        for q in matches:
            try:
                q_num = int(q)

                # Only realistic ISC question numbers
                if 1 <= q_num <= 20:
                    questions.append(q_num)

            except:
                pass

    return sorted(set(questions))


def detect_answer_marking_page(text):
    """
    Detect explicit answer/marking scheme headings.
    This is intentionally strict to avoid false positives.
    """

    patterns = [
        r"\bANSWER\s+KEY\b",
        r"\bANSWERS\b",
        r"\bMARKING\s+SCHEME\b",
        r"\bMARKING\s+SCHEME\s*/\s*ANSWERS\b",
        r"\bANSWERS\s*/\s*MARKING\s+SCHEME\b",
        r"\bSUGGESTED\s+ANSWERS\b",
        r"\bANSWER\s+GUIDELINES\b",
        r"\bEVALUATION\s+GUIDELINES\b"
    ]

    for pattern in patterns:
        if re.search(pattern, text, re.IGNORECASE):
            return True

    return False


def detect_question_section(text):
    """
    Determine whether a page appears to contain question-paper
    content rather than answer-key content.
    """

    question_patterns = [
        r"\bQuestion\s+[1-9]",
        r"\bQ\.?\s*[1-9]",
        r"\bSECTION\s+[AB]\b",
        r"\bAnswer\s+any\b",
        r"\bAttempt\s+any\b"
    ]

    for pattern in question_patterns:
        if re.search(pattern, text, re.IGNORECASE):
            return True

    return False


# ============================================================
# 3. Inspect each PDF page by page
# ============================================================

summary_rows = []
page_rows = []

for file_name in available_files:

    file_path = os.path.join("/content", file_name)

    print("\n" + "=" * 100)
    print("FILE:", file_name)
    print("=" * 100)

    doc = fitz.open(file_path)

    # Extract complete document text
    all_pages_text = []

    for page_number, page in enumerate(doc, start=1):
        text = clean_text(page.get_text())
        all_pages_text.append(text)

    full_text = "\n".join(all_pages_text)

    # --------------------------------------------------------
    # Document-level information
    # --------------------------------------------------------

    document_type = detect_document_type(file_name, full_text)
    year = detect_year(full_text)
    max_marks = detect_max_marks(full_text)

    all_questions = set()
    answer_pages = []
    question_pages = []

    # --------------------------------------------------------
    # Page-level inspection
    # --------------------------------------------------------

    for page_number, page_text in enumerate(all_pages_text, start=1):

        questions = detect_question_numbers(page_text)

        for q in questions:
            all_questions.add(q)

        has_answer_heading = detect_answer_marking_page(page_text)
        has_question_content = detect_question_section(page_text)

        if has_answer_heading:
            answer_pages.append(page_number)

        if has_question_content and not has_answer_heading:
            question_pages.append(page_number)

        page_rows.append({
            "source_file": file_name,
            "page_number": page_number,
            "question_numbers": ", ".join(map(str, questions)),
            "has_question_content": has_question_content,
            "has_answer_marking_heading": has_answer_heading,
            "text_preview": page_text[:500]
        })

    # --------------------------------------------------------
    # Print summary
    # --------------------------------------------------------

    print(f"Pages             : {len(doc)}")
    print(f"Document type     : {document_type}")
    print(f"Detected year     : {year}")
    print(f"Maximum marks     : {max_marks}")
    print(f"Question numbers  : {sorted(all_questions)}")
    print(f"Question pages    : {question_pages}")
    print(f"Answer/mark pages : {answer_pages}")

    # --------------------------------------------------------
    # Show first 2 pages for manual structural confirmation
    # --------------------------------------------------------

    print("\n--- FIRST PAGE PREVIEW ---")

    if len(all_pages_text) > 0:
        print(all_pages_text[0][:1500])

    if len(all_pages_text) > 1:
        print("\n--- SECOND PAGE PREVIEW ---")
        print(all_pages_text[1][:1500])

    summary_rows.append({
        "source_file": file_name,
        "pages": len(doc),
        "document_type": document_type,
        "year": year,
        "maximum_marks": max_marks,
        "question_numbers": ", ".join(map(str, sorted(all_questions))),
        "question_pages": ", ".join(map(str, question_pages)),
        "answer_marking_pages": ", ".join(map(str, answer_pages))
    })

    doc.close()


# ============================================================
# 4. Create summary DataFrames
# ============================================================

df_structure = pd.DataFrame(summary_rows)

df_page_structure = pd.DataFrame(page_rows)


# ============================================================
# 5. Display document-level structure
# ============================================================

print("\n\n")
print("=" * 100)
print("DOCUMENT STRUCTURE SUMMARY")
print("=" * 100)

display(df_structure)


# ============================================================
# 6. Display page-level structure
# ============================================================

print("\n\n")
print("=" * 100)
print("PAGE-LEVEL STRUCTURE")
print("=" * 100)

display(df_page_structure[
    [
        "source_file",
        "page_number",
        "question_numbers",
        "has_question_content",
        "has_answer_marking_heading"
    ]
])


# ============================================================
# 7. Save inspection results
# ============================================================

summary_path = "/content/ISC_Class12_English_PENDING_SOURCE_STRUCTURE.xlsx"

with pd.ExcelWriter(summary_path, engine="openpyxl") as writer:

    df_structure.to_excel(
        writer,
        sheet_name="Document_Summary",
        index=False
    )

    df_page_structure.to_excel(
        writer,
        sheet_name="Page_Structure",
        index=False
    )

print("\n" + "=" * 100)
print("Inspection completed.")
print("Saved:", summary_path)
print("=" * 100)
```

```python
import os
import re
import fitz
import pandas as pd

# ============================================================
# FILE
# ============================================================

file_name = "English Paper 1 - AOPP.pdf"
file_path = os.path.join("/content", file_name)

doc = fitz.open(file_path)

print("=" * 100)
print("EXTRACTING:", file_name)
print("=" * 100)
print("Total pages:", len(doc))


# ============================================================
# EXTRACT ALL PAGES
# ============================================================

pages = {}

for page_no in range(1, len(doc) + 1):
    text = doc[page_no - 1].get_text()

    text = text.replace("\xa0", " ")
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n\s*\n+", "\n\n", text)

    pages[page_no] = text.strip()


# ============================================================
# IMPORTANT:
# Actual English Language paper begins on PAGE 8
# ============================================================

question_pages = range(8, len(doc) + 1)


# ============================================================
# DISPLAY PAGE HEADINGS
# ============================================================

print("\n")
print("=" * 100)
print("PAGE HEADINGS / STRUCTURE")
print("=" * 100)

for page_no in question_pages:

    text = pages[page_no]

    # Show first 700 characters of every relevant page
    print("\n" + "-" * 100)
    print(f"PAGE {page_no}")
    print("-" * 100)
    print(text[:700])


# ============================================================
# FIND QUESTION STARTS
# ============================================================

question_pattern = re.compile(
    r"(?m)^\s*Question\s+([1-9]\d?)\s*$",
    re.IGNORECASE
)

question_locations = []

for page_no in question_pages:

    text = pages[page_no]

    for match in question_pattern.finditer(text):

        q_num = int(match.group(1))

        question_locations.append({
            "question": q_num,
            "page": page_no,
            "position": match.start()
        })


question_locations = pd.DataFrame(question_locations)

print("\n")
print("=" * 100)
print("QUESTION LOCATIONS")
print("=" * 100)

display(question_locations)


# ============================================================
# FIND MARKING SCHEME LOCATIONS
# ============================================================

marking_locations = []

for page_no in question_pages:

    text = pages[page_no]

    matches = re.finditer(
        r"\bMARKING\s+SCHEME\b",
        text,
        re.IGNORECASE
    )

    for match in matches:

        marking_locations.append({
            "page": page_no,
            "position": match.start()
        })


marking_locations = pd.DataFrame(marking_locations)

print("\n")
print("=" * 100)
print("MARKING SCHEME LOCATIONS")
print("=" * 100)

display(marking_locations)


# ============================================================
# EXTRACT QUESTION + MARKING SCHEME BLOCKS
# ============================================================

records = []

for page_no in question_pages:

    text = pages[page_no]

    # --------------------------------------------------------
    # Find all Question X occurrences on this page
    # --------------------------------------------------------

    q_matches = list(
        question_pattern.finditer(text)
    )

    for i, q_match in enumerate(q_matches):

        q_number = int(q_match.group(1))

        start = q_match.start()

        if i + 1 < len(q_matches):
            end = q_matches[i + 1].start()
        else:
            end = len(text)

        block = text[start:end].strip()

        records.append({
            "page": page_no,
            "question_number": q_number,
            "question_block": block
        })


# ============================================================
# CREATE DATAFRAME
# ============================================================

df_aopp_blocks = pd.DataFrame(records)


print("\n")
print("=" * 100)
print("EXTRACTED QUESTION BLOCKS")
print("=" * 100)

display(
    df_aopp_blocks[
        [
            "page",
            "question_number",
            "question_block"
        ]
    ]
)


# ============================================================
# SAVE FOR INSPECTION
# ============================================================

output_file = "/content/ISC_Class12_English_Paper1_AOPP_QUESTION_BLOCKS.xlsx"

df_aopp_blocks.to_excel(
    output_file,
    index=False
)

print("\n")
print("=" * 100)
print("SAVED")
print("=" * 100)
print(output_file)

doc.close()
```

```python
import os
import pandas as pd

# ============================================================
# FILES
# ============================================================

base_path = "/content"

q1_file = os.path.join(
    base_path,
    "ISC_Class12_English_Paper1_AOPP_Q1_EXTRACTED.xlsx"
)

q2_file = os.path.join(
    base_path,
    "ISC_Class12_English_Paper1_AOPP_Q2_EXTRACTED.xlsx"
)

q4_file = os.path.join(
    base_path,
    "ISC_Class12_English_Paper1_AOPP_Q4_DATABASE_READY.xlsx"
)

# ============================================================
# LOAD Q1, Q2, Q4
# ============================================================

df_q1 = pd.read_excel(q1_file)
df_q2 = pd.read_excel(q2_file)
df_q4 = pd.read_excel(q4_file)

# ============================================================
# Q1 — NO FIXED ANSWER PROVIDED
# ============================================================

# AOPP provides evaluation guidance rather than a model answer.
# Therefore answer must remain NULL/None.

if "answer" not in df_q1.columns:
    df_q1["answer"] = None

# ============================================================
# Q3
# ============================================================

q3_records = []

# ------------------------------------------------------------
# Q3(i)
# ------------------------------------------------------------

q3_i = [
    (
        "a",
        "As soon as Mira found her lost diamond pendant, she ran to "
        "her mother in excitement.\n"
        "(Begin: No sooner ………………………………)",
        "No sooner did Mira find her lost diamond pendant than she ran "
        "to her mother in excitement."
    ),
    (
        "b",
        "The doctor said to the elderly lady, "
        "\"Will you be able to follow the schedule for taking medicines "
        "everyday?\"\n"
        "(Begin: The doctor asked ………………………………)",
        "The doctor asked the elderly lady whether/if she would be able "
        "to follow the schedule for taking medicines every day."
    ),
    (
        "c",
        "Heavy rains have caused several landslides.\n"
        "(End: ……………………………… heavy rains.)",
        "Several landslides have been caused by/due to heavy rains."
    ),
    (
        "d",
        "Sunita is the best debater in our team.\n"
        "(Begin: No other ………………………………)",
        "No other debater in our team is as good as Sunita."
    ),
    (
        "e",
        "The news is too good to be true.\n"
        "(Use: …… so……………that……………)",
        "The news is so good that it cannot be true."
    )
]

for letter, question_text, answer in q3_i:
    q3_records.append({
        "parent_question": "Q3",
        "sub_question": "i",
        "question_number": f"Q3(i)({letter})",
        "question_type": "Sentence Transformation",
        "marks": 1,
        "question_text": question_text,
        "answer": answer
    })

# ------------------------------------------------------------
# Q3(ii)
# ------------------------------------------------------------

q3_ii = [
    ("a", "Never put out until tomorrow what you can do today.", "off"),
    ("b", "Put by your best dress. We are going to attend a party.", "on"),
    ("c", "She is so friendly that she gets away with everyone.", "along / on"),
    ("d", "I must get up to writing this assignment as it is due tomorrow.", "down / back"),
    ("e", "The engagement was called through at the last moment.", "off"),
    ("f", "This situation calls about necessary action.", "for"),
    ("g", "Salon chose to stand up her friend in times of crisis.", "by / with"),
    ("h", "IPL stands of Indian Premier League.", "for"),
    ("i", "Neha turned out the job offer as she wanted to continue her higher studies.", "down"),
    ("j", "I did not expect so many people to turn over for the conference.", "up")
]

for letter, question_text, answer in q3_ii:
    q3_records.append({
        "parent_question": "Q3",
        "sub_question": "ii",
        "question_number": f"Q3(ii)({letter})",
        "question_type": "Phrasal Verb",
        "marks": 0.5,
        "question_text": question_text,
        "answer": answer
    })

# ------------------------------------------------------------
# Q3(iii)
# ------------------------------------------------------------

q3_iii = [
    ("1", "Once upon a time, there ________ (1) (live) three old monks.", "lived"),
    ("2", "They ________ (2) (be) poor and without any worldly goods.", "were"),
    ("3", "They would travel from village to village and, in every village they ________ (3) (stop), they would sit in the center of the marketplace and start laughing.", "stopped"),
    ("4", "At first, everyone just ________ (4) (pass) by without giving them any attention.", "passed"),
    ("5", "This was when the monks would finally get up and ________ (5) (move) on to the next village.", "move"),
    ("6", "Their fame ________ (6) (grow) with time.", "grew"),
    ("7", "Nobody ever ________ (7) (preach) the way they did with laughter and nothing else.", "preached / had preached"),
    ("8", "They never laughed at anything in particular, but it was as if they ________ (8) (understand) a cosmic joke.", "understood / had understood"),
    ("9", "Just ________ (9) (imagine)!", "imagine"),
    ("10", "Three monks ________ (10) (spread) joy and laughter without saying a single word.", "spreading / spread")
]

for number, question_text, answer in q3_iii:
    q3_records.append({
        "parent_question": "Q3",
        "sub_question": "iii",
        "question_number": f"Q3(iii)({number})",
        "question_type": "Verb Form",
        "marks": 0.5,
        "question_text": question_text,
        "answer": answer
    })

df_q3 = pd.DataFrame(q3_records)

# ============================================================
# REQUIRED COLUMNS
# ============================================================

required_question_columns = [
    "parent_question",
    "sub_question",
    "question_number",
    "question_type",
    "marks",
    "question_text",
    "answer"
]

# Make sure all four DataFrames have answer column
for df_name, df in [
    ("Q1", df_q1),
    ("Q2", df_q2),
    ("Q3", df_q3),
    ("Q4", df_q4)
]:

    if "answer" not in df.columns:
        df["answer"] = None

    missing = [
        col for col in required_question_columns
        if col not in df.columns
    ]

    if missing:
        raise ValueError(
            f"{df_name} is missing columns: {missing}"
        )

# ============================================================
# COMBINE
# ============================================================

df_questions = pd.concat(
    [
        df_q1[required_question_columns],
        df_q2[required_question_columns],
        df_q3[required_question_columns],
        df_q4[required_question_columns]
    ],
    ignore_index=True
)

# ============================================================
# COMMON METADATA
# ============================================================

df_questions["board"] = "ISC"
df_questions["class"] = "Class 12"
df_questions["subject"] = "English"
df_questions["year"] = 2025
df_questions["paper"] = "Paper 1 - English Language"
df_questions["source_type"] = "CISCE AOPP"
df_questions["source_file"] = "English Paper 1 - AOPP.pdf"

# ============================================================
# QUESTION IDs
# ============================================================

df_questions["question_id"] = (
    "ISC12_ENG1_AOPP_2025_"
    + df_questions["question_number"]
    .str.replace("(", "_", regex=False)
    .str.replace(")", "", regex=False)
    .str.replace("-", "_", regex=False)
)

# ============================================================
# INSTRUCTIONS
# ============================================================

def get_instruction(row):

    qn = row["question_number"]

    if qn.startswith("Q1"):
        return (
            "Write a composition in approximately 400–450 words "
            "on any one of the given subjects."
        )

    if qn == "Q2(i)":
        return (
            "Write a feature article in about 300 words using the "
            "given guidelines."
        )

    if qn == "Q2(ii)":
        return (
            "Write a proposal in about 150 words stating the steps "
            "you would take to organise the programme and make it a success."
        )

    if qn.startswith("Q3(i)"):
        return (
            "Rewrite the following sentences according to the instructions "
            "given after each of them in brackets."
        )

    if qn.startswith("Q3(ii)"):
        return (
            "Replace the incorrect underlined words to make the sentences "
            "meaningful and grammatically correct. Do not copy the sentences."
        )

    if qn.startswith("Q3(iii)"):
        return (
            "Fill in the blanks with the appropriate form of the verb "
            "given in brackets."
        )

    if qn.startswith("Q4(i)(a)"):
        return (
            "Find a single word from the passage that will exactly replace "
            "the underlined word or words."
        )

    if qn.startswith("Q4(i)(b)"):
        return (
            "Choose the correct sentence that uses the same word unchanged "
            "in spelling, but with a different meaning from that in the passage."
        )

    if qn.startswith("Q4(ii)"):
        return (
            "Answer the following questions as briefly as possible in your own words."
        )

    if qn == "Q4(iii)":
        return (
            "Summarise the given passage in the form of a connected passage "
            "in about 100 words."
        )

    return None

df_questions["instruction"] = df_questions.apply(
    get_instruction,
    axis=1
)

# ============================================================
# FINAL 16 COLUMNS
# ============================================================

final_columns = [
    "question_id",
    "board",
    "class",
    "subject",
    "year",
    "paper",
    "source_type",
    "parent_question",
    "sub_question",
    "question_type",
    "marks",
    "question_text",
    "answer",
    "source_file",
    "question_number",
    "instruction"
]

df_final = df_questions[final_columns].copy()

# ============================================================
# VALIDATION
# ============================================================

print("=" * 100)
print("FINAL AOPP DATASET VALIDATION")
print("=" * 100)

print("\nTotal records:", len(df_final))
print("Expected: 43")

print("\nRecords by parent question:")
print(
    df_final.groupby("parent_question")
    .size()
    .to_string()
)

print("\nStored marks by parent:")
print(
    df_final.groupby("parent_question")["marks"]
    .sum()
    .to_string()
)

# Q1 alternatives = 20 actual marks, not 120
actual_marks = (
    20
    + df_final.loc[
        df_final["parent_question"] == "Q2", "marks"
    ].sum()
    + df_final.loc[
        df_final["parent_question"] == "Q3", "marks"
    ].sum()
    + df_final.loc[
        df_final["parent_question"] == "Q4", "marks"
    ].sum()
)

print("\nActual paper marks:", actual_marks)
print("Expected: 80")

# Duplicate IDs
duplicates = df_final[
    df_final["question_id"].duplicated(keep=False)
]

print("\nDuplicate IDs:", len(duplicates))

# Missing question text
print(
    "Missing question text:",
    df_final["question_text"].isna().sum()
)

# Missing answers
print(
    "Missing answers:",
    df_final["answer"].isna().sum()
)

# Q1 answers should be NULL
print(
    "Q1 records with NULL answer:",
    df_final.loc[
        df_final["parent_question"] == "Q1",
        "answer"
    ].isna().sum()
)

print("\nQuestion types:")
print(
    df_final["question_type"]
    .value_counts()
    .to_string()
)

print("\nNumber of columns:", len(df_final.columns))
print("Expected: 16")

# ============================================================
# SAVE
# ============================================================

csv_file = (
    "/content/"
    "ISC_Class12_English_Paper1_AOPP_2025_DATABASE_READY.csv"
)

xlsx_file = (
    "/content/"
    "ISC_Class12_English_Paper1_AOPP_2025_DATABASE_READY.xlsx"
)

df_final.to_csv(
    csv_file,
    index=False,
    encoding="utf-8-sig"
)

df_final.to_excel(
    xlsx_file,
    index=False
)

print("\nFiles saved:")
print(csv_file)
print(xlsx_file)

# ============================================================
# PREVIEW Q1
# ============================================================

print("\n")
print("=" * 100)
print("Q1 ANSWER CHECK")
print("=" * 100)

display(
    df_final[
        df_final["parent_question"] == "Q1"
    ][
        [
            "question_id",
            "question_number",
            "question_type",
            "marks",
            "answer"
        ]
    ]
)
```

```python
# ============================================================
# GENERATE MYSQL INSERT SCRIPT
# ISC CLASS 12 ENGLISH PAPER 1 - AOPP 2025
# ============================================================

import pandas as pd
import math

# ------------------------------------------------------------
# INPUT
# ------------------------------------------------------------

csv_file = "/content/ISC_Class12_English_Paper1_AOPP_2025_DATABASE_READY.csv"

df_sql = pd.read_csv(
    csv_file,
    encoding="utf-8-sig"
)

# ------------------------------------------------------------
# COLUMN ORDER
# ------------------------------------------------------------

columns = [
    "question_id",
    "board",
    "class",
    "subject",
    "year",
    "paper",
    "source_type",
    "parent_question",
    "sub_question",
    "question_type",
    "marks",
    "question_text",
    "answer",
    "source_file",
    "question_number",
    "instruction"
]

df_sql = df_sql[columns]

# ------------------------------------------------------------
# HELPER FUNCTION
# ------------------------------------------------------------

def mysql_value(value):
    """
    Convert a Python/Pandas value into a MySQL-safe SQL value.

    Text is stored as UTF-8 hexadecimal using:
        CONVERT(0x... USING utf8mb4)

    This avoids Unicode/encoding problems in MySQL Workbench.
    """

    if pd.isna(value):
        return "NULL"

    # Integer year
    if isinstance(value, (int,)):
        return str(value)

    # Float marks
    if isinstance(value, float):
        if math.isnan(value):
            return "NULL"
        return str(value)

    # Handle numpy numeric types
    if hasattr(value, "item"):
        try:
            value = value.item()
            if isinstance(value, (int, float)):
                if isinstance(value, float) and math.isnan(value):
                    return "NULL"
                return str(value)
        except Exception:
            pass

    # Text
    text = str(value)

    utf8_hex = text.encode("utf-8").hex()

    return f"CONVERT(0x{utf8_hex} USING utf8mb4)"


# ------------------------------------------------------------
# CREATE SQL
# ------------------------------------------------------------

sql_lines = []

sql_lines.append("-- =====================================================")
sql_lines.append("-- ISC CLASS 12 ENGLISH PAPER 1 - AOPP 2025")
sql_lines.append("-- DATABASE INSERT SCRIPT")
sql_lines.append("-- =====================================================")
sql_lines.append("")
sql_lines.append("USE isc_assessment;")
sql_lines.append("")
sql_lines.append("-- Existing records with these IDs will be replaced.")
sql_lines.append("")

# Delete existing AOPP records first
sql_lines.append(
    "DELETE FROM questions "
    "WHERE board='ISC' "
    "AND class='Class 12' "
    "AND subject='English' "
    "AND year=2025 "
    "AND paper='Paper 1 - English Language' "
    "AND source_type='CISCE AOPP';"
)

sql_lines.append("")

# ------------------------------------------------------------
# INSERT STATEMENTS
# ------------------------------------------------------------

column_string = ", ".join(columns)

for _, row in df_sql.iterrows():

    values = [
        mysql_value(row[col])
        for col in columns
    ]

    value_string = ", ".join(values)

    sql = (
        f"INSERT INTO questions ({column_string}) "
        f"VALUES ({value_string});"
    )

    sql_lines.append(sql)

# ------------------------------------------------------------
# VALIDATION QUERIES
# ------------------------------------------------------------

sql_lines.append("")
sql_lines.append("-- =====================================================")
sql_lines.append("-- VALIDATION")
sql_lines.append("-- =====================================================")
sql_lines.append("")

sql_lines.append("""
SELECT
    COUNT(*) AS total_records
FROM questions
WHERE board='ISC'
  AND class='Class 12'
  AND subject='English'
  AND year=2025
  AND paper='Paper 1 - English Language'
  AND source_type='CISCE AOPP';
""".strip())

sql_lines.append("")

sql_lines.append("""
SELECT
    parent_question,
    COUNT(*) AS record_count,
    SUM(marks) AS stored_marks
FROM questions
WHERE board='ISC'
  AND class='Class 12'
  AND subject='English'
  AND year=2025
  AND paper='Paper 1 - English Language'
  AND source_type='CISCE AOPP'
GROUP BY parent_question
ORDER BY parent_question;
""".strip())

sql_lines.append("")

sql_lines.append("""
SELECT
    COUNT(*) AS missing_question_text
FROM questions
WHERE board='ISC'
  AND class='Class 12'
  AND subject='English'
  AND year=2025
  AND paper='Paper 1 - English Language'
  AND source_type='CISCE AOPP'
  AND (question_text IS NULL OR TRIM(question_text) = '');
""".strip())

sql_lines.append("")

sql_lines.append("""
SELECT
    COUNT(*) AS missing_answers
FROM questions
WHERE board='ISC'
  AND class='Class 12'
  AND subject='English'
  AND year=2025
  AND paper='Paper 1 - English Language'
  AND source_type='CISCE AOPP'
  AND answer IS NULL;
""".strip())

# ------------------------------------------------------------
# SAVE SQL FILE
# ------------------------------------------------------------

sql_file = (
    "/content/"
    "ISC_Class12_English_Paper1_AOPP_2025_INSERT.sql"
)

with open(sql_file, "w", encoding="utf-8") as f:
    f.write("\n".join(sql_lines))

# ------------------------------------------------------------
# RESULT
# ------------------------------------------------------------

print("=" * 100)
print("MYSQL SCRIPT GENERATED")
print("=" * 100)

print("\nRecords in SQL:", len(df_sql))
print("SQL file:")
print(sql_file)

print("\nFile exists:", __import__("os").path.exists(sql_file))

print("\nExpected database result:")
print("Total AOPP records: 43")
print("Q1: 6")
print("Q2: 2")
print("Q3: 25")
print("Q4: 10")
print("Missing answers: 6")
```

```python
# ============================================================
# ISC CLASS 12 ENGLISH PAPER 1 - SPECIMEN 2025
# FINAL CONSOLIDATION: Q1 + Q2 + Q3 + Q4
# ============================================================

import pandas as pd

# ------------------------------------------------------------
# 1. COMBINE ALL FOUR SECTIONS
# ------------------------------------------------------------

df_final = pd.concat(
    [df_q1, df_q2, df_q3, df_q4],
    ignore_index=True
)

print("=" * 100)
print("SPECIMEN PAPER 1 - FINAL CONSOLIDATION")
print("=" * 100)

print("\nTotal records:", len(df_final))
print("Expected records: 43")


# ------------------------------------------------------------
# 2. ADD METADATA
# ------------------------------------------------------------

df_final["board"] = "ISC"
df_final["class"] = "Class 12"
df_final["subject"] = "English"
df_final["year"] = 2025
df_final["paper"] = "Paper 1 - English Language"
df_final["source_type"] = "CISCE Specimen"
df_final["source_file"] = "English Paper 1 - Specimen.pdf"


# ------------------------------------------------------------
# 3. CREATE QUESTION IDs
# ------------------------------------------------------------

def create_question_id(row):

    q = row["parent_question"]
    sub = row["sub_question"]
    qno = row["question_number"]

    # Q1 / Q2
    if q in ["Q1", "Q2"]:
        return f"ISC12_ENG1_SPEC_2025_{q}_{sub}"

    # Q3
    elif q == "Q3":
        # Example: Q3 + i + a
        parts = qno.replace("Q3(", "").replace(")", "").split("(")

        if len(parts) >= 2:
            return f"ISC12_ENG1_SPEC_2025_Q3_{parts[0]}_{parts[1]}"
        else:
            return f"ISC12_ENG1_SPEC_2025_Q3_{sub}"

    # Q4
    elif q == "Q4":
        parts = qno.replace("Q4(", "").replace(")", "").split("(")

        if len(parts) == 3:
            return f"ISC12_ENG1_SPEC_2025_Q4_{parts[0]}_{parts[1]}_{parts[2]}"
        elif len(parts) == 2:
            return f"ISC12_ENG1_SPEC_2025_Q4_{parts[0]}_{parts[1]}"
        else:
            return f"ISC12_ENG1_SPEC_2025_Q4_{sub}"

    return None


df_final["question_id"] = df_final.apply(create_question_id, axis=1)


# ------------------------------------------------------------
# 4. ARRANGE EXACT DATABASE COLUMNS
# ------------------------------------------------------------

columns = [
    "question_id",
    "board",
    "class",
    "subject",
    "year",
    "paper",
    "source_type",
    "parent_question",
    "sub_question",
    "question_type",
    "marks",
    "question_text",
    "answer",
    "source_file",
    "question_number",
    "instruction"
]

df_final = df_final[columns]


# ------------------------------------------------------------
# 5. VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("VALIDATION")
print("=" * 100)

# Record count
print("\n1. Total records:", len(df_final))
print("   Expected: 43")

# Column count
print("\n2. Number of columns:", len(df_final.columns))
print("   Expected: 16")

# Duplicate IDs
duplicate_ids = df_final["question_id"].duplicated().sum()

print("\n3. Duplicate question IDs:", duplicate_ids)
print("   Expected: 0")

# Missing question text
missing_question_text = df_final["question_text"].isna().sum()

print("\n4. Missing question text:", missing_question_text)
print("   Expected: 0")

# Missing answers
missing_answers = df_final["answer"].isna().sum()

print("\n5. Missing answers:", missing_answers)
print("   Expected: 0")

# Section distribution
print("\n6. Records by section:")
print(df_final["parent_question"].value_counts().sort_index())

# Marks by section
print("\n7. Stored marks by section:")
print(
    df_final.groupby("parent_question")["marks"]
    .sum()
    .sort_index()
)

# Question type distribution
print("\n8. Question type distribution:")
print(df_final["question_type"].value_counts())

# Source metadata
print("\n9. Source metadata:")
print("Board:", df_final["board"].unique())
print("Class:", df_final["class"].unique())
print("Subject:", df_final["subject"].unique())
print("Year:", df_final["year"].unique())
print("Paper:", df_final["paper"].unique())
print("Source type:", df_final["source_type"].unique())

# Actual paper marks
actual_paper_marks = 20 + 25 + 15 + 20

print("\n10. Actual paper-level marks:", actual_paper_marks)
print("    Expected: 80")


# ------------------------------------------------------------
# 6. CHECK QUESTION IDs
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("QUESTION ID SAMPLE")
print("=" * 100)

display(
    df_final[
        [
            "question_id",
            "parent_question",
            "sub_question",
            "question_number"
        ]
    ].head(15)
)


# ------------------------------------------------------------
# 7. FINAL DATASET PREVIEW
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("FINAL DATASET")
print("=" * 100)

display(df_final)


# ------------------------------------------------------------
# 8. FINAL PASS / FAIL CHECK
# ------------------------------------------------------------

checks = {
    "Record count = 43": len(df_final) == 43,
    "Column count = 16": len(df_final.columns) == 16,
    "No duplicate IDs": duplicate_ids == 0,
    "No missing question text": missing_question_text == 0,
    "No missing answers": missing_answers == 0,
    "Q1 = 6 records": len(df_final[df_final["parent_question"] == "Q1"]) == 6,
    "Q2 = 2 records": len(df_final[df_final["parent_question"] == "Q2"]) == 2,
    "Q3 = 25 records": len(df_final[df_final["parent_question"] == "Q3"]) == 25,
    "Q4 = 10 records": len(df_final[df_final["parent_question"] == "Q4"]) == 10,
    "Actual paper marks = 80": actual_paper_marks == 80
}

print("\n" + "=" * 100)
print("FINAL VALIDATION RESULT")
print("=" * 100)

all_passed = True

for check, result in checks.items():
    status = "PASS" if result else "FAIL"
    print(f"{status}: {check}")

    if not result:
        all_passed = False

print("\n" + "=" * 100)

if all_passed:
    print("ALL VALIDATION CHECKS PASSED")
else:
    print("SOME VALIDATION CHECKS FAILED - DO NOT EXPORT YET")
```

```python
# ============================================================
# ISC CLASS 12 ENGLISH PAPER 1 - VEDANTU 2025
# DATABASE-READY EXPORT
# ============================================================

import pandas as pd
import os

print("=" * 100)
print("VEDANTU PAPER 1 - DATABASE-READY EXPORT")
print("=" * 100)

# ------------------------------------------------------------
# FILE NAMES
# ------------------------------------------------------------

base_name = "ISC_Class12_English_Paper1_Vedantu_2025_DATABASE_READY"

csv_file = f"/content/{base_name}.csv"
xlsx_file = f"/content/{base_name}.xlsx"
sql_file = f"/content/{base_name}_INSERT.sql"


# ------------------------------------------------------------
# 1. FINAL DATAFRAME CHECK
# ------------------------------------------------------------

assert len(df_final) == 51, "Record count is not 51"
assert len(df_final.columns) == 16, "Column count is not 16"
assert df_final["question_id"].duplicated().sum() == 0, \
    "Duplicate question IDs found"

print("\nFinal dataframe verified:")
print("Records:", len(df_final))
print("Columns:", len(df_final.columns))


# ------------------------------------------------------------
# 2. EXPORT CSV
# ------------------------------------------------------------

df_final.to_csv(
    csv_file,
    index=False,
    encoding="utf-8-sig"
)

print("\nCSV exported:")
print(csv_file)


# ------------------------------------------------------------
# 3. EXPORT EXCEL
# ------------------------------------------------------------

df_final.to_excel(
    xlsx_file,
    index=False,
    engine="openpyxl"
)

print("\nExcel exported:")
print(xlsx_file)


# ------------------------------------------------------------
# 4. GENERATE MYSQL INSERT SCRIPT
# ------------------------------------------------------------

def mysql_value(value):

    # NULL
    if pd.isna(value):
        return "NULL"

    # Numeric values
    if isinstance(value, (int, float)) and not isinstance(value, bool):
        return str(value)

    # String values
    value = str(value)

    # Escape MySQL special characters
    value = value.replace("\\", "\\\\")
    value = value.replace("'", "''")
    value = value.replace("\n", "\\n")
    value = value.replace("\r", "\\r")

    return "'" + value + "'"


with open(sql_file, "w", encoding="utf-8") as f:

    f.write("-- ======================================================\n")
    f.write("-- ISC Class 12 English Paper 1 - Vedantu 2025\n")
    f.write("-- Database: isc_assessment\n")
    f.write("-- Table: questions\n")
    f.write("-- Records: 51\n")
    f.write("-- ======================================================\n\n")

    f.write("USE isc_assessment;\n\n")

    for _, row in df_final.iterrows():

        columns_sql = ", ".join(df_final.columns)

        values_sql = ", ".join(
            mysql_value(row[col])
            for col in df_final.columns
        )

        sql = (
            f"INSERT INTO questions ({columns_sql}) "
            f"VALUES ({values_sql});\n"
        )

        f.write(sql)


print("\nSQL exported:")
print(sql_file)


# ------------------------------------------------------------
# 5. FILE SIZE CHECK
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("EXPORTED FILES")
print("=" * 100)

for file_path in [csv_file, xlsx_file, sql_file]:

    size_kb = os.path.getsize(file_path) / 1024

    print(
        f"{os.path.basename(file_path):65s} "
        f"{size_kb:.2f} KB"
    )


# ------------------------------------------------------------
# 6. PREVIEW
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("DATABASE-READY PREVIEW")
print("=" * 100)

display(
    df_final[
        [
            "question_id",
            "parent_question",
            "sub_question",
            "question_type",
            "marks",
            "question_number"
        ]
    ].head(15)
)

print("\n" + "=" * 100)
print("EXPORT COMPLETED")
print("=" * 100)

print("\nFiles created successfully.")
print("Next step: check MySQL for existing Vedantu Paper 1 records")
print("before running the INSERT script.")
```

```python
# ============================================================
# ISC CLASS 12 ENGLISH
# COMPETENCY-FOCUSED PRACTICE QUESTIONS
# DATABASE-READY DATASET
# Q1 - Q89
# ============================================================

import pandas as pd
import os
import re


# ============================================================
# 1. FILE PATHS
# ============================================================

question_file = (
    "/content/"
    "ISC_Class12_English_Competency_Practice_ALL_QUESTIONS.txt"
)

answer_file = (
    "/content/"
    "ISC_Class12_English_Competency_Practice_ALL_ANSWERS.txt"
)

source_file = (
    "/content/"
    "English Paper - Competency Focused Practice question.pdf"
)

if not os.path.exists(question_file):
    raise FileNotFoundError(question_file)

if not os.path.exists(answer_file):
    raise FileNotFoundError(answer_file)

print("Question extraction found.")
print("Answer extraction found.")


# ============================================================
# 2. READ EXTRACTED TEXT
# ============================================================

with open(question_file, "r", encoding="utf-8") as f:
    question_text = f.read()

with open(answer_file, "r", encoding="utf-8") as f:
    answer_text = f.read()


# ============================================================
# 3. EXTRACT QUESTION BLOCKS
# ============================================================

# We use question numbers at the beginning of a line.
# Example:
# 1.
# 2.
# ...
# 89.

question_pattern = re.compile(
    r"(?ms)^\s*(\d{1,2})\.\s+(.*?)(?=^\s*\d{1,2}\.\s+|\Z)"
)

question_matches = question_pattern.findall(question_text)

question_dict = {}

for number, text in question_matches:

    number = int(number)

    if 1 <= number <= 89:
        question_dict[number] = text.strip()


print("\n" + "=" * 100)
print("QUESTION EXTRACTION")
print("=" * 100)

print("Questions detected:", len(question_dict))

missing_questions = [
    q for q in range(1, 90)
    if q not in question_dict
]

print("Missing questions:", missing_questions)

assert len(question_dict) == 89, (
    f"Expected 89 questions, found {len(question_dict)}"
)

assert len(missing_questions) == 0, (
    f"Missing question numbers: {missing_questions}"
)


# ============================================================
# 4. EXTRACT ANSWER BLOCKS
# ============================================================

# The answer key uses:
# 1.
# 2.
# ...
# 89.
#
# We extract each answer until the next numbered answer.

answer_pattern = re.compile(
    r"(?ms)^\s*(\d{1,2})\.\s+(.*?)(?=^\s*\d{1,2}\.\s+|\Z)"
)

answer_matches = answer_pattern.findall(answer_text)

answer_dict = {}

for number, text in answer_matches:

    number = int(number)

    if 1 <= number <= 89:
        answer_dict[number] = text.strip()


print("\n" + "=" * 100)
print("ANSWER EXTRACTION")
print("=" * 100)

print("Answers detected:", len(answer_dict))

missing_answers = [
    q for q in range(1, 90)
    if q not in answer_dict
]

print("Questions without extracted answer:", missing_answers)


# ============================================================
# IMPORTANT:
# We do NOT invent answers.
#
# If an answer is genuinely absent from the source,
# it will remain NULL.
# ============================================================


# ============================================================
# 5. SECTION INFORMATION
# ============================================================

def get_section(question_number):

    if 1 <= question_number <= 20:
        return "MCQ"

    elif 21 <= question_number <= 55:
        return "Complete the Sentences"

    elif 56 <= question_number <= 73:
        return "Short Answer"

    elif 74 <= question_number <= 89:
        return "Long Answer"

    return None


def get_marks(question_number):

    if 1 <= question_number <= 55:
        return 1

    elif 56 <= question_number <= 73:
        return 5

    elif 74 <= question_number <= 89:
        return 10

    return None


def get_instruction(question_number):

    if 1 <= question_number <= 20:
        return "Choose the correct option."

    elif 21 <= question_number <= 55:
        return "Complete the sentence as instructed."

    elif 56 <= question_number <= 73:
        return "Answer in about 150 words, incorporating the details specified in the question."

    elif 74 <= question_number <= 89:
        return "Answer in about 200-250 words, or as specified in the question."

    return ""


def get_question_type(question_number):

    if 1 <= question_number <= 20:
        return "MCQ"

    elif 21 <= question_number <= 55:
        return "Complete the Sentence"

    elif 56 <= question_number <= 73:
        return "Short Answer"

    elif 74 <= question_number <= 89:
        return "Long Answer"

    return None


# ============================================================
# 6. BUILD DATABASE RECORDS
# ============================================================

records = []

for q_no in range(1, 90):

    section = get_section(q_no)

    record = {
        "question_id":
            f"ISC12_ENG_CFP_2025_Q{q_no}",

        "board":
            "ISC",

        "class":
            "Class 12",

        "subject":
            "English",

        "year":
            2025,

        "paper":
            "Competency-Focused Practice - Literature in English",

        "source_type":
            "CISCE Competency-Focused Practice",

        "parent_question":
            f"Q{q_no}",

        "sub_question":
            None,

        "question_type":
            get_question_type(q_no),

        "marks":
            get_marks(q_no),

        "question_text":
            question_dict.get(q_no),

        "answer":
            answer_dict.get(q_no),

        "source_file":
            "English Paper - Competency Focused Practice question.pdf",

        "question_number":
            str(q_no),

        "instruction":
            get_instruction(q_no)
    }

    records.append(record)


# ============================================================
# 7. CREATE DATAFRAME
# ============================================================

df = pd.DataFrame(records)


print("\n" + "=" * 100)
print("DATAFRAME CREATED")
print("=" * 100)

print("Rows    :", len(df))
print("Columns :", len(df.columns))

display(df.head(10))


# ============================================================
# 8. COLUMN VALIDATION
# ============================================================

expected_columns = [
    "question_id",
    "board",
    "class",
    "subject",
    "year",
    "paper",
    "source_type",
    "parent_question",
    "sub_question",
    "question_type",
    "marks",
    "question_text",
    "answer",
    "source_file",
    "question_number",
    "instruction"
]

assert list(df.columns) == expected_columns

print("\nColumn validation: PASSED")


# ============================================================
# 9. RECORD COUNT VALIDATION
# ============================================================

assert len(df) == 89

print("Record count: 89 PASSED")


# ============================================================
# 10. DUPLICATE ID CHECK
# ============================================================

duplicate_ids = df[
    df["question_id"].duplicated(keep=False)
]

print("\nDuplicate IDs:", len(duplicate_ids))

assert len(duplicate_ids) == 0


# ============================================================
# 11. MISSING QUESTION TEXT CHECK
# ============================================================

missing_question_text = df[
    df["question_text"].isna() |
    (df["question_text"].astype(str).str.strip() == "")
]

print("Missing question text:", len(missing_question_text))

assert len(missing_question_text) == 0


# ============================================================
# 12. MISSING ANSWER CHECK
# ============================================================

missing_answer = df[
    df["answer"].isna() |
    (df["answer"].astype(str).str.strip() == "")
]

print("Missing answers:", len(missing_answer))

if len(missing_answer) > 0:

    print("\nQuestions with missing answers:")
    display(
        missing_answer[
            [
                "question_number",
                "question_type",
                "question_text"
            ]
        ]
    )


# ============================================================
# 13. METADATA VALIDATION
# ============================================================

assert df["board"].eq("ISC").all()
assert df["class"].eq("Class 12").all()
assert df["subject"].eq("English").all()
assert df["year"].eq(2025).all()

print("\nMetadata validation: PASSED")


# ============================================================
# 14. SECTION VALIDATION
# ============================================================

section_summary = (
    df.groupby("question_type")
      .agg(
          record_count=("question_id", "count"),
          stored_marks=("marks", "sum")
      )
      .reset_index()
)

print("\n" + "=" * 100)
print("SECTION SUMMARY")
print("=" * 100)

display(section_summary)


# Expected:
#
# MCQ                    20 records   20 marks
# Complete the Sentence  35 records   35 marks
# Short Answer           18 records   90 marks
# Long Answer            16 records  160 marks


expected_counts = {
    "MCQ": 20,
    "Complete the Sentence": 35,
    "Short Answer": 18,
    "Long Answer": 16
}

for qtype, expected in expected_counts.items():

    actual = int(
        df.loc[
            df["question_type"] == qtype,
            "question_id"
        ].count()
    )

    assert actual == expected, (
        f"{qtype}: expected {expected}, found {actual}"
    )


print("Section counts: PASSED")


# ============================================================
# 15. MARKS VALIDATION
# ============================================================

expected_marks = {
    "MCQ": 20,
    "Complete the Sentence": 35,
    "Short Answer": 90,
    "Long Answer": 160
}

for qtype, expected in expected_marks.items():

    actual = int(
        df.loc[
            df["question_type"] == qtype,
            "marks"
        ].sum()
    )

    assert actual == expected, (
        f"{qtype}: expected {expected} marks, found {actual}"
    )


total_stored_marks = int(df["marks"].sum())

print("\nTotal question-level marks:", total_stored_marks)

assert total_stored_marks == 305

print("Marks validation: PASSED")


# ============================================================
# IMPORTANT NOTE
# ============================================================

print("\nNOTE:")
print(
    "305 is the sum of the source-defined marks across all 89 "
    "practice questions. This document is a competency-focused "
    "question bank, NOT a single 305-mark examination paper."
)


# ============================================================
# 16. QUESTION NUMBER RANGE CHECK
# ============================================================

expected_numbers = list(range(1, 90))

actual_numbers = (
    df["question_number"]
    .astype(int)
    .tolist()
)

assert actual_numbers == expected_numbers

print("\nQuestion numbering 1-89: PASSED")


# ============================================================
# 17. QUESTION TYPE DISTRIBUTION
# ============================================================

print("\n" + "=" * 100)
print("QUESTION TYPE DISTRIBUTION")
print("=" * 100)

display(
    df["question_type"]
    .value_counts()
    .rename_axis("question_type")
    .reset_index(name="record_count")
)


# ============================================================
# 18. SHOW LAST QUESTIONS
# ============================================================

print("\n" + "=" * 100)
print("Q73-Q89 CHECK")
print("=" * 100)

display(
    df[
        df["question_number"].astype(int) >= 73
    ][
        [
            "question_number",
            "question_type",
            "marks",
            "question_text",
            "answer"
        ]
    ]
)


# ============================================================
# 19. EXPORT CSV
# ============================================================

csv_path = (
    "/content/"
    "ISC_Class12_English_Competency_Practice_2025_DATABASE_READY.csv"
)

df.to_csv(
    csv_path,
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 20. EXPORT EXCEL
# ============================================================

xlsx_path = (
    "/content/"
    "ISC_Class12_English_Competency_Practice_2025_DATABASE_READY.xlsx"
)

df.to_excel(
    xlsx_path,
    index=False,
    engine="openpyxl"
)


# ============================================================
# 21. GENERATE MYSQL INSERT SQL
# ============================================================

sql_path = (
    "/content/"
    "ISC_Class12_English_Competency_Practice_2025_INSERT.sql"
)


def sql_value(value):

    if pd.isna(value):
        return "NULL"

    value = str(value)

    # Escape backslash
    value = value.replace("\\", "\\\\")

    # Escape single quote
    value = value.replace("'", "''")

    return "'" + value + "'"


sql_columns = [
    "question_id",
    "board",
    "`class`",
    "subject",
    "year",
    "paper",
    "source_type",
    "parent_question",
    "sub_question",
    "question_type",
    "marks",
    "question_text",
    "answer",
    "source_file",
    "question_number",
    "instruction"
]


with open(sql_path, "w", encoding="utf-8") as f:

    f.write("USE isc_assessment;\n\n")

    for _, row in df.iterrows():

        values = [
            sql_value(row["question_id"]),
            sql_value(row["board"]),
            sql_value(row["class"]),
            sql_value(row["subject"]),
            str(int(row["year"])),
            sql_value(row["paper"]),
            sql_value(row["source_type"]),
            sql_value(row["parent_question"]),
            sql_value(row["sub_question"]),
            sql_value(row["question_type"]),
            str(int(row["marks"])),
            sql_value(row["question_text"]),
            sql_value(row["answer"]),
            sql_value(row["source_file"]),
            sql_value(row["question_number"]),
            sql_value(row["instruction"])
        ]

        sql = (
            "INSERT INTO questions ("
            + ", ".join(sql_columns)
            + ") VALUES ("
            + ", ".join(values)
            + ");\n"
        )

        f.write(sql)


# ============================================================
# 22. FINAL VALIDATION SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("FINAL VALIDATION SUMMARY")
print("=" * 100)

print("Total records       :", len(df))
print("Total columns       :", len(df.columns))
print("Duplicate IDs       :", df["question_id"].duplicated().sum())
print("Missing questions   :", len(missing_question_text))
print("Missing answers     :", len(missing_answer))
print("Total stored marks  :", total_stored_marks)

print("\nCSV:")
print(csv_path)

print("\nExcel:")
print(xlsx_path)

print("\nMySQL SQL:")
print(sql_path)

print("\n" + "=" * 100)
print("DATASET BUILD COMPLETED")
print("=" * 100)
```
